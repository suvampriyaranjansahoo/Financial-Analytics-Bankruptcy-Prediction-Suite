# 07B – SHAP Local Explainability (Updated)

Uses the **cleaned dataset** and **production model**.

```
import pandas as pd
import joblib
import shap

DATA_PATH='american_bankruptcy_cleaned.csv'
MODEL_PATH='production_bankruptcy_model.joblib'

df=pd.read_csv(DATA_PATH)
target='status_label' if 'status_label' in df.columns else 'target'
X=df.drop(columns=[target])

model=joblib.load(MODEL_PATH)
instance=X.iloc[[0]]

explainer=shap.TreeExplainer(model)
values=explainer(instance)
shap.plots.waterfall(values[0],show=False)
```
