# 06F - CatBoost Classifier

Professional notebook for training and evaluating a CatBoost model.

## 1. Import Libraries

```
import pandas as pd
import joblib
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, classification_report,
    ConfusionMatrixDisplay, RocCurveDisplay,
    PrecisionRecallDisplay
)
from catboost import CatBoostClassifier
```

## 2. Load Dataset

```
df=pd.read_csv("american_bankruptcy.csv")
df["target"]=df["status_label"].map({"alive":0,"failed":1})
display(df.head())
```

## 3. Prepare Features

```
drop_cols=["status_label","target"]
if "company_name" in df.columns:
    drop_cols.append("company_name")
X=df.drop(columns=drop_cols)
y=df["target"]

num=X.select_dtypes(include="number").columns
cat=X.select_dtypes(exclude="number").columns
```

## 4. Train/Test Split

```
X_train,X_test,y_train,y_test=train_test_split(
    X,y,test_size=0.2,stratify=y,random_state=42)
```

## 5. Build Preprocessing Pipeline

```
preprocessor=ColumnTransformer([
("num",SimpleImputer(strategy="median"),num),
("cat","passthrough",cat)
])
```

## 6. Initialize CatBoost

```
model=Pipeline([
("preprocessor",preprocessor),
("classifier",CatBoostClassifier(
    iterations=300,
    learning_rate=0.05,
    depth=6,
    loss_function="Logloss",
    eval_metric="AUC",
    verbose=0,
    random_seed=42
))
])
```

## 7. Train Model

```
model.fit(X_train,y_train)
```

## 8. Cross Validation

```
scores=cross_val_score(model,X_train,y_train,cv=5,scoring="f1")
print(scores)
print("Mean F1:",scores.mean())
```

## 9. Predictions

```
pred=model.predict(X_test)
proba=model.predict_proba(X_test)[:,1]
```

## 10. Evaluation Metrics

```
results=pd.DataFrame({
"Metric":["Accuracy","Precision","Recall","F1","ROC-AUC"],
"Score":[
accuracy_score(y_test,pred),
precision_score(y_test,pred),
recall_score(y_test,pred),
f1_score(y_test,pred),
roc_auc_score(y_test,proba)]
})
display(results)
print(classification_report(y_test,pred))
```

## 11. Confusion Matrix

```
ConfusionMatrixDisplay.from_predictions(y_test,pred)
plt.show()
```

## 12. ROC Curve

```
RocCurveDisplay.from_predictions(y_test,proba)
plt.show()
```

## 13. Precision-Recall Curve

```
PrecisionRecallDisplay.from_predictions(y_test,proba)
plt.show()
```

## 14. Feature Importance

```
clf=model.named_steps["classifier"]
imp=pd.DataFrame({
    "Feature":num.tolist()+cat.tolist(),
    "Importance":clf.feature_importances_[:len(num)+len(cat)]
})
display(imp.sort_values("Importance",ascending=False).head(20))
```

## 15. Save Model

```
joblib.dump(model,"catboost_model.joblib")
print("Saved catboost_model.joblib")
```
