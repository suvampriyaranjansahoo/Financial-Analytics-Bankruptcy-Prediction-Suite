# 05D - Feature Scaling & Transformation

Apply common scaling techniques and distribution transformations while following ML best practices.

```python

import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler

df = pd.read_csv("../data/datasets/american_bankruptcy.csv")
df['target'] = df['status_label'].map({'alive':0,'failed':1})

drop_cols=['status_label','target']
if 'company_name' in df.columns:
    drop_cols.append('company_name')

X=df.drop(columns=drop_cols)
numeric=X.select_dtypes(include='number').columns
display(X.head())
```

## Standard Scaling

```python

std_scaler=StandardScaler()
X_std=X.copy()
X_std[numeric]=std_scaler.fit_transform(X[numeric])
display(X_std.head())
```

## Min-Max Scaling

```python

mm_scaler=MinMaxScaler()
X_mm=X.copy()
X_mm[numeric]=mm_scaler.fit_transform(X[numeric])
display(X_mm.head())
```

## Robust Scaling

```python

rb_scaler=RobustScaler()
X_rb=X.copy()
X_rb[numeric]=rb_scaler.fit_transform(X[numeric])
display(X_rb.head())
```

## Log Transformation

```python

X_log=X.copy()
created=[]
for c in numeric:
    if (X_log[c]>=0).all():
        X_log[c]=np.log1p(X_log[c])
        created.append(c)
print("Log transformed columns:",created)
display(X_log.head())
```

## Scaling Comparison

```python

summary=pd.DataFrame({
'Original Mean':X[numeric].mean(),
'Standard Mean':X_std[numeric].mean().round(3),
'MinMax Min':X_mm[numeric].min().round(3),
'MinMax Max':X_mm[numeric].max().round(3)
})
display(summary.head(10))
```

## Export Scaled Datasets

```python

X_std.to_csv("scaled_standard.csv",index=False)
X_mm.to_csv("scaled_minmax.csv",index=False)
X_rb.to_csv("scaled_robust.csv",index=False)
print("Saved:")
print("- scaled_standard.csv")
print("- scaled_minmax.csv")
print("- scaled_robust.csv")
```
