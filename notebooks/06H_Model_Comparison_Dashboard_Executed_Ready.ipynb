# 06H - Model Comparison Dashboard (Execution-Ready)

Compare all trained models using a consolidated dashboard.

```
import pandas as pd
import matplotlib.pyplot as plt

# Replace these values with your actual results after running notebooks 06A-06G
results=pd.DataFrame({
'Model':['Logistic Regression','Decision Tree','Random Forest','XGBoost','LightGBM','CatBoost','SVM'],
'Accuracy':[0,0,0,0,0,0,0],
'Precision':[0,0,0,0,0,0,0],
'Recall':[0,0,0,0,0,0,0],
'F1':[0,0,0,0,0,0,0],
'ROC_AUC':[0,0,0,0,0,0,0]
})
display(results)
```

## Best Model by Metric

```
for metric in ['Accuracy','Precision','Recall','F1','ROC_AUC']:
    best=results.loc[results[metric].idxmax()]
    print(f'{metric}: {best["Model"]} -> {best[metric]}')
```

## Metric Comparison Charts

```
for metric in ['Accuracy','Precision','Recall','F1','ROC_AUC']:
    plt.figure(figsize=(8,4))
    plt.bar(results['Model'],results[metric])
    plt.xticks(rotation=30,ha='right')
    plt.title(metric)
    plt.tight_layout()
    plt.show()
```

## Ranking

```
ranked=results.copy()
ranked['Overall Score']=ranked[['Accuracy','Precision','Recall','F1','ROC_AUC']].mean(axis=1)
ranked=ranked.sort_values('Overall Score',ascending=False)
display(ranked)
```

## Business Recommendation

```
best=ranked.iloc[0]
print(f'Recommended production model: {best["Model"]}')
print('Recommendation is based on average performance across all evaluation metrics.')
```
