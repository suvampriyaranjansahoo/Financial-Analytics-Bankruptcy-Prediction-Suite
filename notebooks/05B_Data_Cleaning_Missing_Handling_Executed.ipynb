# 05B - Data Cleaning & Missing Value Handling

Enterprise preprocessing notebook for cleaning the dataset prior to feature engineering.

```python
import pandas as pd
import numpy as np

df=pd.read_csv(r"/mnt/data/american_bankruptcy.csv")
```

## Dataset Snapshot

```python
display(df.head())
print(df.shape)
```

## Missing Value Analysis

```python
missing=df.isna().sum().to_frame('Missing Count')
missing['Missing %']=(missing['Missing Count']/len(df)*100).round(2)
display(missing.sort_values('Missing Count',ascending=False))
```

## Duplicate Checks

```python
print('Full duplicate rows:',df.duplicated().sum())
if {'company_name','year'}.issubset(df.columns):
    print('Duplicate company-year:',df.duplicated(['company_name','year']).sum())
```

## Data Type Validation

```python
display(df.dtypes.to_frame('Data Type'))
num=df.select_dtypes(include='number')
print('Numeric columns:',len(num.columns))
```

## Constant & Near-Constant Features

```python
const=[c for c in df.columns if df[c].nunique()==1]
print('Constant features:',const)
near=df.nunique().sort_values()
display(near.to_frame('Unique Values').head(10))
```

## Basic Cleaning

```python
clean=df.copy()
clean.columns=clean.columns.str.strip()
if clean.duplicated().sum():
    clean=clean.drop_duplicates()
print('Shape after cleaning:',clean.shape)
```

## Data Quality Report

```python
report=pd.DataFrame({
'Check':['Missing Values','Duplicate Rows','Constant Features'],
'Result':[df.isna().sum().sum(),df.duplicated().sum(),len(const)]
})
display(report)
```

## Save Clean Dataset

```python
clean.to_csv('clean_dataset.csv',index=False)
print('Saved: clean_dataset.csv')
```
