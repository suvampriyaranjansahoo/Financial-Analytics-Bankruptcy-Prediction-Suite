```python

import pandas as pd
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize']=(10,5)
df=pd.read_csv(r"/mnt/data/american_bankruptcy.csv")
df.head()
```

## Dataset Shape

```python
print(df.shape)
```

## Dataset Info

```python
df.info()
```

## Summary Statistics

```python
df.describe(include='all').T
```

## Missing Values

```python
df.isna().sum().to_frame('Missing')
```

## Duplicate Rows

```python
print(df.duplicated().sum())
```

## Target Distribution

```python

vc=df['status_label'].value_counts()
print(vc)
vc.plot(kind='bar',title='Target Distribution')
plt.show()
```

## Year Distribution

```python

yr=df['year'].value_counts().sort_index()
yr.plot(marker='o',title='Records by Year')
plt.show()
```

## Company Overview

```python

print("Unique companies:",df['company_name'].nunique())
df.groupby('company_name').size().describe()
```
