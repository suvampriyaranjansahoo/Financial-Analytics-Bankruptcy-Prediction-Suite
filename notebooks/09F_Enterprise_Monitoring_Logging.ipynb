# 09F – Enterprise Monitoring & Logging

Implement monitoring, logging, and basic model observability for the deployed Bankruptcy Risk Prediction platform.

## Business Objective

Track API usage, prediction quality, latency, failures, and system health so the deployed model can be operated reliably in production.

```python
import logging
import time
import pandas as pd
from pathlib import Path

LOG_DIR = Path("logs")
LOG_DIR.mkdir(exist_ok=True)

logging.basicConfig(
    filename=LOG_DIR / "prediction_service.log",
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s"
)
```

```python
def log_prediction(features, prediction, probability, latency_ms):
    logging.info(
        f"prediction={prediction} "
        f"probability={probability:.4f} "
        f"latency_ms={latency_ms:.2f}"
    )

sample_metrics = pd.DataFrame({
    "Metric":[
        "Total Requests",
        "Successful Predictions",
        "Failed Requests",
        "Average Latency (ms)",
        "Model Version"
    ],
    "Value":[1500,1492,8,34.7,"1.0.0"]
})

sample_metrics.to_csv("monitoring_metrics.csv",index=False)
sample_metrics
```
