# 09D – Streamlit Web Application (Enterprise)

Create an interactive web application for bankruptcy risk prediction.

## Business Objective
Provide business users with a simple interface to submit company financial features and receive real-time bankruptcy risk predictions.

````
import streamlit as st
import pandas as pd
import joblib

MODEL_PATH='production_bankruptcy_model.joblib'

model=joblib.load(MODEL_PATH)
df=pd.read_csv('american_bankruptcy_cleaned.csv')

target='status_label' if 'status_label' in df.columns else 'target'
features=[c for c in df.columns if c!=target]

st.set_page_config(page_title='Bankruptcy Risk Prediction',layout='wide')
st.title('🏦 Bankruptcy Risk Prediction Dashboard')
st.write('Enter company financial values to predict bankruptcy risk.')
````

````
inputs={}
col1,col2=st.columns(2)

for i,feature in enumerate(features):
    default=float(df[feature].median())
    if i % 2 == 0:
        with col1:
            inputs[feature]=st.number_input(feature,value=default)
    else:
        with col2:
            inputs[feature]=st.number_input(feature,value=default)
````

````
if st.button('Predict Bankruptcy Risk'):
    X=pd.DataFrame([inputs])
    pred=int(model.predict(X)[0])
    prob=float(model.predict_proba(X)[0][1])

    if pred==1:
        st.error(f'High Bankruptcy Risk ({prob:.2%})')
    else:
        st.success(f'Low Bankruptcy Risk ({prob:.2%})')

    st.metric('Prediction','Bankrupt' if pred else 'Healthy')
    st.progress(min(max(prob,0.0),1.0))
    st.dataframe(X)
````

## Features

- Interactive input form
- Real-time predictions
- Bankruptcy probability
- Business-friendly output
- Ready for deployment with Streamlit Cloud or Docker
