# 07D – Executive Explainability Report

Enterprise-level report consolidating SHAP and LIME explainability results for business stakeholders.

## Business Objective
Summarize the explainability analysis, identify the main drivers of bankruptcy risk, and provide actionable recommendations for executives, risk analysts, auditors, and investors.

```
import pandas as pd
from pathlib import Path

# Load explanation outputs if available
global_path = Path("shap_global_feature_importance.csv")
local_path = Path("local_shap_explanation.csv")
lime_path = Path("lime_feature_contributions.csv")

global_df = pd.read_csv(global_path) if global_path.exists() else None
local_df = pd.read_csv(local_path) if local_path.exists() else None
lime_df = pd.read_csv(lime_path) if lime_path.exists() else None

print("Global SHAP:", global_df is not None)
print("Local SHAP:", local_df is not None)
print("LIME:", lime_df is not None)
```

## Executive Summary

This report combines:

- Global SHAP Explainability (overall model behavior)
- Local SHAP Explainability (single prediction)
- LIME Explainability (model-agnostic local explanation)

Together, these methods improve transparency and trust in the bankruptcy prediction model.

## Key Findings

### Global SHAP
- Identifies the financial variables with the greatest influence across all predictions.
- Helps determine the strongest bankruptcy indicators.

### Local SHAP
- Explains why a specific company received its bankruptcy prediction.
- Quantifies the contribution of each feature.

### LIME
- Provides an independent local explanation.
- Useful for validating SHAP findings and communicating results to non-technical stakeholders.


## Business Recommendations

### Credit Risk Teams
- Investigate companies with multiple high-risk financial indicators.
- Use feature importance to prioritize manual reviews.

### Investors
- Monitor companies whose key financial ratios consistently increase bankruptcy risk.

### Auditors
- Focus on firms exhibiting combinations of the most influential risk factors.

### Executives
- Deploy explainability dashboards to support transparent and defensible AI decisions.
