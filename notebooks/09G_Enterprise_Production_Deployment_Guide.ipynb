# 09G – Enterprise Production Deployment Guide

Deployment documentation for the complete Bankruptcy Risk Prediction platform.

## Business Objective

Provide a repeatable deployment guide covering local execution, Docker, cloud deployment, verification, maintenance, and troubleshooting.

````
# Local Development

# 1. Create environment
python -m venv .venv

# 2. Activate
# Windows:
# .venv\Scripts\activate
# Linux/Mac:
# source .venv/bin/activate

# 3. Install
pip install -r requirements.txt

# 4. Start API
uvicorn api.app:app --reload

# 5. Start Streamlit
streamlit run streamlit_app/app.py
````

## Verification Checklist

- API reachable (`/health`)
- Swagger available (`/docs`)
- Streamlit dashboard loads
- Model loads successfully
- Single prediction works
- Batch prediction works
- SHAP explanations render
- Logs are generated
- Monitoring metrics update


## Troubleshooting

| Issue | Resolution |
|-------|------------|
| Model not loading | Verify model path and joblib version |
| Missing packages | Reinstall requirements.txt |
| Port conflict | Change API/Streamlit ports |
| Docker build fails | Rebuild without cache |
| Prediction errors | Validate feature schema |
