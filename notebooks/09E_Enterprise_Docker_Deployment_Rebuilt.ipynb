# 09E – Enterprise Docker & Container Deployment

Package the complete enterprise application (FastAPI + Multi-page Streamlit + Model) for production deployment.

## Business Objective

Containerize the analytics platform so it can be deployed consistently on local machines, Render, Azure Container Apps, AWS ECS, Google Cloud Run, or Kubernetes.

```
# Project Structure

Bankruptcy-Risk-Prediction/
│
├── api/
│   └── app.py
├── streamlit_app/
│   ├── app.py
│   ├── pages/
│   ├── assets/
│   ├── css/
│   └── utils/
├── models/
│   ├── production_bankruptcy_model.joblib
│   ├── model_metadata.json
│   └── model_schema.csv
└── Dockerfile

```

```dockerfile
# Dockerfile

FROM python:3.11-slim

WORKDIR /app

COPY . .

RUN pip install --no-cache-dir -r requirements.txt

EXPOSE 8000
EXPOSE 8501

CMD ["uvicorn","api.app:app","--host","0.0.0.0","--port","8000"]

```

```yaml
# docker-compose.yml

version: "3.9"

services:

  api:
    build: .
    container_name: bankruptcy-api
    command: uvicorn api.app:app --host 0.0.0.0 --port 8000
    ports:
      - "8000:8000"

  streamlit:
    build: .
    container_name: bankruptcy-dashboard
    command: streamlit run streamlit_app/app.py --server.port=8501 --server.address=0.0.0.0
    depends_on:
      - api
    ports:
      - "8501:8501"

```

```text
# requirements.txt (core)

fastapi
uvicorn
streamlit
pandas
numpy
scikit-learn
joblib
plotly
shap
lime
matplotlib

```
