````
# app.py (main)

import streamlit as st

st.set_page_config(
    page_title='Enterprise Bankruptcy Analytics',
    page_icon='📊',
    layout='wide',
    initial_sidebar_state='expanded'
)

st.logo('assets/logo.png')

st.sidebar.success('Navigation')
st.title('Enterprise Bankruptcy Risk Analytics Platform')

st.markdown('''
### Welcome

Use the sidebar to explore:

- Executive Dashboard
- Single Prediction
- Batch Prediction
- Explainability
- Model Performance
- Reports
''')
````
