# Bankruptcy Risk Prediction
## Phase 04 - Data Cleaning & Preprocessing

This notebook creates:
- `american_bankruptcy_cleaned.csv` (analysis-ready)
- `american_bankruptcy_model_ready.csv` (ML-ready)


```
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings("ignore")

DATA_PATH="american_bankruptcy.csv"
df=pd.read_csv(DATA_PATH)
print(df.shape)
df.head()
```

## Dataset Quality Report

```
print(df.info())
print("\nMissing Values")
print(df.isnull().sum())
print("\nDuplicates:",df.duplicated().sum())
```

## Missing Value Treatment

```
num=df.select_dtypes(include=np.number).columns
cat=df.select_dtypes(exclude=np.number).columns

for c in num:
    df[c]=df[c].fillna(df[c].median())

for c in cat:
    df[c]=df[c].fillna(df[c].mode()[0])
```

## Remove Duplicates

```
before=len(df)
df=df.drop_duplicates()
print("Removed",before-len(df),"duplicates")
```

## Clean Text Columns

```
for c in df.select_dtypes(include="object").columns:
    df[c]=df[c].astype(str).str.strip()
```

## Save Analysis Dataset

```
df.to_csv("american_bankruptcy_cleaned.csv",index=False)
print("Saved american_bankruptcy_cleaned.csv")
```

## Create Model-ready Dataset

```
model_df=df.copy()

if "company_name" in model_df.columns:
    model_df=model_df.drop(columns=["company_name"])

target=None
for c in ["status_label","status","bankrupt"]:
    if c in model_df.columns:
        target=c
        break

if target:
    model_df[target]=model_df[target].astype(str).str.lower().map({
        "alive":0,
        "failed":1,
        "bankrupt":1,
        "non-bankrupt":0
    }).fillna(model_df[target])

model_df.to_csv("american_bankruptcy_model_ready.csv",index=False)

print(model_df.dtypes)
print("Saved american_bankruptcy_model_ready.csv")
```

## Final Validation

```
print("Analysis Dataset:",df.shape)
print("Model Dataset:",model_df.shape)
print("Cleaning Completed Successfully.")
```
