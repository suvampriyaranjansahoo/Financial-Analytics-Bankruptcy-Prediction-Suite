# 06H v2 – Professional Model Comparison Dashboard
Automatically compare all trained models and generate a production recommendation.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ===========================
# UPDATE THESE VALUES AFTER RUNNING 06A-06G
# ===========================
results = pd.DataFrame({
    "Model":["Logistic Regression","Decision Tree","Random Forest","XGBoost","LightGBM","CatBoost","SVM"],
    "Accuracy":[0,0,0,0,0,0,0],
    "Precision":[0,0,0,0,0,0,0],
    "Recall":[0,0,0,0,0,0,0],
    "F1":[0,0,0,0,0,0,0],
    "ROC_AUC":[0,0,0,0,0,0,0],
    "Training_Time_sec":[0,0,0,0,0,0,0],
    "Inference_ms":[0,0,0,0,0,0,0],
    "Model_Size_MB":[0,0,0,0,0,0,0]
})
display(results)
```

## Metric Leaderboard

```
metrics=["Accuracy","Precision","Recall","F1","ROC_AUC"]
for m in metrics:
    best=results.loc[results[m].idxmax()]
    print(f"{m}: {best['Model']} ({best[m]:.4f})")
```

## Overall Weighted Score

```
weights={
    "Accuracy":0.20,
    "Precision":0.20,
    "Recall":0.20,
    "F1":0.20,
    "ROC_AUC":0.20
}
results["Overall_Score"]=sum(results[k]*v for k,v in weights.items())
leaderboard=results.sort_values("Overall_Score",ascending=False)
display(leaderboard)
```

## Performance Charts

```
for metric in ["Accuracy","Precision","Recall","F1","ROC_AUC"]:
    plt.figure(figsize=(8,4))
    plt.bar(results["Model"],results[metric])
    plt.xticks(rotation=30,ha="right")
    plt.title(metric)
    plt.tight_layout()
    plt.show()
```

## Efficiency Comparison

```
figsize=(8,4)
for metric in ["Training_Time_sec","Inference_ms","Model_Size_MB"]:
    plt.figure(figsize=figsize)
    plt.bar(results["Model"],results[metric])
    plt.xticks(rotation=30,ha="right")
    plt.title(metric)
    plt.tight_layout()
    plt.show()
```

## Production Recommendation

```
best=leaderboard.iloc[0]

print("="*60)
print("FINAL PRODUCTION RECOMMENDATION")
print("="*60)
print(f"Selected Model : {best['Model']}")
print(f"Overall Score  : {best['Overall_Score']:.4f}")
print("\nReason:")
print("- Best overall predictive performance")
print("- Balanced classification metrics")
print("- Suitable for deployment after hyperparameter tuning")
```
