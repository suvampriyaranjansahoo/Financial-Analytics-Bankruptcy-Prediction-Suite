# 07A – SHAP Global Explainability (Updated)

Uses the **cleaned dataset** and the saved **production model**.

```python
import pandas as pd
import joblib
import shap
import matplotlib.pyplot as plt

DATA_PATH='../data/datasets/american_bankruptcy_cleaned.csv'
MODEL_PATH='../models/production_bankruptcy_model.joblib'

df=pd.read_csv(DATA_PATH)
target='status_label' if 'status_label' in df.columns else 'target'
X=df.drop(columns=[target])
if 'company_name' in X.columns:
    X=X.drop(columns=['company_name'])

model=joblib.load(MODEL_PATH)

sample=X.sample(min(500,len(X)),random_state=42)
explainer=shap.TreeExplainer(model)
shap_values=explainer.shap_values(sample)

shap.summary_plot(shap_values,sample,show=False)
plt.tight_layout()
plt.savefig('shap_summary.png',dpi=300)

shap.summary_plot(shap_values,sample,plot_type='bar',show=False)
plt.tight_layout()
plt.savefig('shap_bar.png',dpi=300)
```

**Outputs:** `shap_summary.png`, `shap_bar.png`.
