# 04 - Business Insights & Executive Report

Final notebook for Phase 1 summarizing analytical findings and business recommendations.

```python
import pandas as pd
df=pd.read_csv("../data/datasets/american_bankruptcy.csv")
```

## Executive Summary

```python
summary=pd.DataFrame({
'Metric':['Total Records','Companies','Years Covered','Alive','Failed'],
'Value':[len(df),
df['company_name'].nunique(),
f"{df['year'].min()}-{df['year'].max()}",
(df['status_label']=='alive').sum(),
(df['status_label']!='alive').sum()]
})
display(summary)
```

## Key Findings

```python
findings=[
'The dataset spans multiple fiscal years, enabling longitudinal analysis.',
'The target variable is imbalanced, requiring appropriate evaluation metrics.',
'Financial variables contain outliers that should be reviewed before preprocessing.',
'Some variables are correlated and may require feature selection.'
]
for i,f in enumerate(findings,1):
    print(f'{i}. {f}')
```

## Business Implications

```python
implications=[
'Early identification of financially distressed firms can improve lending and investment decisions.',
'Risk models should prioritize Recall, Precision, F1-score, ROC-AUC and PR-AUC.',
'Historical trends can support proactive monitoring strategies.'
]
for i,x in enumerate(implications,1):
    print(f'{i}. {x}')
```
